```python
# ============================================================
# IMPORTACIONES
# ============================================================
import os
import pandas as pd
from langchain_mistralai import ChatMistralAI
from langchain_community.embeddings import HuggingFaceEmbeddings
from langchain_community.vectorstores import FAISS
from langchain.text_splitter import CharacterTextSplitter
from langchain.schema import Document
from langchain.chains import RetrievalQA
from langchain.prompts import PromptTemplate

print("✅ Importaciones realizadas correctamente")
```

```python
# ============================================================
# CONFIGURACIÓN DEL API KEY DE MISTRAL
# ============================================================
MISTRAL_API_KEY = "QxpqzC8TTViLaDlJeIrYk29vEsHRx2vI"
os.environ["MISTRAL_API_KEY"] = MISTRAL_API_KEY
print("✅ API Key configurada correctamente")
```

```python
# ============================================================
# CONSTRUCCIÓN DEL CORPUS Y VECTORSTORE (FAISS)
# ============================================================

# Cargar el dataset ORIGINAL (con valores reales, no normalizados)
df_original = pd.read_csv('/content/ventas.csv.csv', encoding='latin1')
df_original.columns = df_original.columns.str.strip().str.lower()

print(f"Dataset cargado: {df_original.shape[0]} filas × {df_original.shape[1]} columnas")
print(f"Columnas: {list(df_original.columns)}")

# Convertir cada fila en un documento de texto para el corpus
def fila_a_texto(row):
    partes = []
    for col, val in row.items():
        partes.append(f"{col}: {val}")
    return " | ".join(partes)

documentos = []
for _, row in df_original.iterrows():
    texto = fila_a_texto(row)
    documentos.append(Document(page_content=texto, metadata={"fuente": "ventas.csv"}))

print(f"✅ Corpus creado: {len(documentos)} documentos")

# Crear embeddings locales (sin necesidad de API)
print("⏳ Creando embeddings (esto puede tardar 1-2 min la primera vez)...")
embeddings = HuggingFaceEmbeddings(model_name="sentence-transformers/all-MiniLM-L6-v2")

# Construir vectorstore FAISS
vectorstore = FAISS.from_documents(documentos, embeddings)
retriever = vectorstore.as_retriever(search_kwargs={"k": 10})

print("✅ Vectorstore FAISS creado correctamente")
print(f"   Documentos indexados: {vectorstore.index.ntotal}")
```

```python
# ============================================================
# CREACIÓN DEL AGENTE LANGCHAIN + MISTRAL (API: ccp)
# ============================================================

# Instanciar el LLM de Mistral via LangChain
llm = ChatMistralAI(
    api_key=MISTRAL_API_KEY,
    model="mistral-small-latest",
    temperature=0
)

# Prompt personalizado para análisis de ventas
prompt_template = PromptTemplate(
    input_variables=["context", "question"],
    template="""Eres un agente analista de ventas experto.
Usa únicamente los datos del contexto siguiente para responder la pregunta.
Si no puedes responder con los datos proporcionados, indícalo claramente.
Responde siempre en español, de forma concisa y directa.

Contexto (registros del dataset de ventas):
{context}

Pregunta: {question}

Respuesta:"""
)

# Cadena de recuperación + generación (RAG)
cadena_qa = RetrievalQA.from_chain_type(
    llm=llm,
    chain_type="stuff",
    retriever=retriever,
    chain_type_kwargs={"prompt": prompt_template},
    return_source_documents=False
)

print("✅ Agente LangChain creado correctamente")
print("   LLM  : mistral-small-latest (API: ccp)")
print("   RAG  : FAISS + sentence-transformers/all-MiniLM-L6-v2")
print(f"   Corpus: {df_original.shape[0]} registros indexados")
```

```python
# ============================================================
# SESIÓN DE PREGUNTAS AL AGENTE (basadas en el dataset real)
# ============================================================
import textwrap

SEP = "=" * 60

def preguntar(pregunta, numero, emoji):
    print(f"\n{SEP}")
    print(f"  {emoji} PREGUNTA {numero}")
    print(SEP)
    for linea in textwrap.wrap(pregunta, width=56):
        print(f"  {linea}")
    print(f"\n  🤖 RESPUESTA DEL AGENTE:")
    print("-" * 60)
    try:
        resultado = cadena_qa.invoke({"query": pregunta})
        respuesta = resultado["result"]
        for linea in respuesta.split('\n'):
            if linea.strip():
                for sub in textwrap.wrap(linea.strip(), width=56):
                    print(f"  {sub}")
            else:
                print()
    except Exception as e:
        print(f"  ⚠️ Error: {e}")
    print(SEP)

print(SEP)
print("  🧠 AGENTE LANGCHAIN + MISTRAL  |  Corpus: ventas.csv")
print(SEP)

# ── Pregunta 1 ───────────────────────────────────────────────
preguntar(
    "¿Cuál es la línea de productos (productline) que aparece con mayor frecuencia en el dataset?",
    1, "📦"
)

# ── Pregunta 2 ───────────────────────────────────────────────
preguntar(
    "¿Cuál es el número de pedido (ordernumber) con la mayor cantidad ordenada (quantityordered)? ¿Cuántas unidades tiene?",
    2, "🔢"
)

# ── Pregunta 3 ───────────────────────────────────────────────
preguntar(
    "¿Cuál es el estado (status) más común en los pedidos del dataset?",
    3, "📋"
)

# ── Pregunta 4 ───────────────────────────────────────────────
preguntar(
    "¿Cuál es el nombre del cliente (customername) con las ventas (sales) más altas registradas?",
    4, "🏆"
)

print()
print(SEP)
print("  ✅  Sesión completada  |  4 preguntas procesadas")
print(SEP)
```
